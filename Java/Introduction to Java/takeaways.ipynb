# Terminology

- class: a container for organizing Java code
- comment: human readable information in Java code that Java ignores
    - Single line: `//`
    - Multi line: `/* .... */`
- method: like sub-containers for code inside of classes that is re-usable
- main method: the starting point or kick off place for a Java application
- Java is case sensitive
- Uses camelcase
    - upper camel case for class names
    - lower camel case for method names
- Each statement ends with a semi-colon
- Does not care about white spaces
- Each block is defined by braces: `{..}`
- NOTE : Variables defined inside a codeblock is unavailable outside that codeblock
- Variables are named/labelled holders of data
- Java is a "type safe" or "strongly typed" language
- The most basic types in Java are called primitive types
- There are 8 primitive types:
    - byte
    - short
    - int
    - long
    - float
    - double
    - boolean
    - char
- Casting: Type casting or just casting converts one data type to another
    - Casting from a smaller type to a larger is done automatically by Java (Example: turning a short into a long)
    - Narrowing casting converts a larger type into a smaller type and required explicit expression (Example: turning a double into a float)
- String: Strings are a collection of characters between double quotes. (Non-primitive type that uses character arrays)
- Arrays : An array is a sequence of data values of fixed size of homogeneous type (It has 0-based indexing)
- Methods : Allows to execute re-usable blocks of code
    - Other than `void`, methods of all other types return some type of value
- Control Flow Conditionals: The `if-else if-else` block in java that work based on conditionals.
    - Alternative to it is `switch..case`.
- Control Flow Loop: The `while`, `do..while`, `for` and `for each` loops based on conditionals
    - Make sure the condition in a loop is false at some point. Otherwise, it will be infinite loop.
- Exception : Occurs due to interruption of normal control flow in an application. 
    - Checked exceptions : That happen during compile time
    - Un-Checked exceptions : That happen during run time
- Exception handling: Allows us to address the issue and continue normal execution of the application (recover the control flow)
    - Use `try..catch...finally` to handle exception
    - `finally` is used for writing teardown/cleanup code since this block is always executed 
- Error: Which cannot be recovered from (example: out or memory)

```python

// Casting
// upcasting
short balance = 100;
long longBalance = balance; //longBalance is 100 but now as a long
// downcasting to narrow type
double y = 5.5d; // double uses 64 bits of memory
float x = (float) 5.5d; // float only uses 32 bits of memory
```

# Methods

```python
public class HelloWorld {
public static void main (String[] args) {
    double smallTip = computeTip(55.0, 0.20); //compute a 20 percent tip on 55 dollars, Calling the method
    double bigTip = computeTip(135.50, 0.33); //compute a 33 percent tip on 135 dollars, Calling the method
    System.out.println("Small tip: $" + smallTip);
    System.out.println("Nice tip: $" + bigTip);
}

// Define the method
static double computeTip (double billAmount, double tipPercentage) {
        double tip = billAmount * tipPercentage;
        return tip;
    }
}
```

# Control Flow (`If-Else`) with Logical Expressions

```python
int a = 14;

if (a > 10) && (a < 20) {
    System.out.println ("Size is average");
} else if (a == 10) || (a == 20) {
    System.out.println ("Size is on the border");
} else {
    System.out.println ("Size is abnormal");
}
```

# Switch : Alternative to If-Else

```python
char direction = 'Z';
switch (direction) {
    case 'E':
        System.out.println("Go east");
        break;
    case 'W':
        System.out.println("Go west");
        break;
    case 'N':
        System.out.println("Go north");
        break;
    case 'S':
        System.out.println("Go south");
        break;
    default: // else alternative that works in switch
        System.out.println("We are lost");
}
```

# For loop

```python
// For loop
for (int i = 0; i < 10; i++){
    System.out.println("Loop number: " + i);
}

// For each loop (normally used array to take from the beginning to the end)
int[] prices = {6, 4, 3, 7};
int sum = 0;
for(int price: prices){
    sum = sum + price;
}
System.out.println(sum);
```

# While Loop

```python
// While loop
int i = 5;
while (i < 100) {
    i = i * 2;
    System.out.println("Looping ...");
}

// Do-While loop (Even if the conditions are not met, the loop will be executed at least once)
int i = 5;
do {
    i = i * 2;
    System.out.println("Looping...");
} while (i < 100);
```

# Exception Handling

```python
int x = 10;
int y = 0;
try {
    System.out.println(x/y);
} catch (ArithmeticException ex){
    System.out.println("Divide by zero?");
} catch (Exception e){ // chaining multiple exceptions
    System.out.println("Some other exception");
} finally {
    System.out.println("Clean up here");
}
```
