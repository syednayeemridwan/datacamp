# Terminology

- class: a container for organizing Java code
- comment: human readable information in Java code that Java ignores
    - Single line: `//`
    - Multi line: `/* .... */`
- method: like sub-containers for code inside of classes that is re-usable
- main method: the starting point or kick off place for a Java application
- Java is case sensitive
- Uses camelcase
    - upper camel case for class names
    - lower camel case for method names
- Each statement ends with a semi-colon
- Does not care about white spaces
- Each block is defined by braces: `{..}`
- Variables are named/labelled holders of data
- Java is a "type safe" or "strongly typed" language
- The most basic types in Java are called primitive types
- There are 8 primitive types:
    - byte
    - short
    - int
    - long
    - float
    - double
    - boolean
    - char
- Casting: Type casting or just casting converts one data type to another
    - Casting from a smaller type to a larger is done automatically by Java (Example: turning a short into a long)
    - Narrowing casting converts a larger type into a smaller type and required explicit expression (Example: turning a double into a float)
- String: Strings are a collection of characters between double quotes. (Non-primitive type that uses character arrays)
- Arrays : An array is a sequence of data values of fixed size of homogeneous type (It has 0-based indexing)
- Methods : Allows to execute re-usable blocks of code
    - Other than `void`, methods of all other types return some type of value

```python

// Casting
// upcasting
short balance = 100;
long longBalance = balance; //longBalance is 100 but now as a long
// downcasting to narrow type
double y = 5.5d; // double uses 64 bits of memory
float x = (float) 5.5d; // float only uses 32 bits of memory
```

# Methods

```python
public class HelloWorld {
public static void main (String[] args) {
    double smallTip = computeTip(55.0, 0.20); //compute a 20 percent tip on 55 dollars, Calling the method
    double bigTip = computeTip(135.50, 0.33); //compute a 33 percent tip on 135 dollars, Calling the method
    System.out.println("Small tip: $" + smallTip);
    System.out.println("Nice tip: $" + bigTip);
}

// Define the method
static double computeTip (double billAmount, double tipPercentage) {
        double tip = billAmount * tipPercentage;
        return tip;
    }
}
```
