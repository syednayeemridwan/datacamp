# Following Conditional Control Flow

Following the logic in multiple and nested conditional (if-then) statements can be challenging. But it is a necessary skill, especially when working with code that is not yours. Given the following Java code, what is the resulting output?

```python
int temperature = 68;
int heatingPoint = 70;
int coolingPoint = 75;
boolean isFanOn = false;
if ((temperature >= heatingPoint) && (temperature <= coolingPoint)) {
    if (isFanOn) {
        System.out.println("Turning fan off");
        isFanOn = false;
    }
} else if ((temperature > coolingPoint) && !isFanOn) {
    System.out.println("Turning on air conditioner and fan");
    isFanOn = true;
} else if ((temperature < heatingPoint) && !isFanOn) {
    System.out.println("Turning on the heat and fan");
    isFanOn = true;
}
```

- Turning on the heat and fan

You now know how to have Java make decisions - that is conditionally execute code based on conditions - using if/then/else. In this exercise, you write a program that determines a student's grade from a score and prints the appropriate message based on the grade all using conditionals.

```python
class ReportCard {

    public static void main(String[] args) {
        int score = 85;
        char grade = assignGrade(score);
        // Add the condition to check if a person's grade is F
		if (grade == 'F') {
            System.out.println("Sorry, you did not pass");
        } // Add and else statement to print a passing score when the grade is not F
        else {
            System.out.println("You have passed the course with a grade of: " + grade);
        }
    }
    
    public static char assignGrade (int score){
        // Add a conditiional to check if the score parameter is larger than 90
        if (score > 90) {
            return 'A';
        } // Add an else conditional to check if the score is larger than 80
        else if (score > 80) {
            return 'B';
        } else if (score > 70) {
            return 'C';
        }
        return 'F';
    }
}
```

# Using a while loop

While loops are good for looping until some condition is met but you aren't sure when that condition will be met - or how many iterations you need to make. For example, while loops are good for searching until something you are looking for is found. In this exercise, you search a String for a specific character. When its found, you print the index of where the character was found in the String. This exercise also gives you a chance to see another conditional and the not (!) operator.

```python
class SearchAString {
    public static void main(String[] args) {
        String message = "It's a good day";
        char searchingFor = 'd';
        boolean found = false;
        int index = 0;
        // Add a while loop to continue until not found
        while (!found) {
            // Check if the character at the current index equals the searched character
            if (message.charAt(index) == searchingFor){
                // if the searchingFor character is located set found to true
                found = true;
                System.out.println("The character was found at index: " + index);
            } else {
                // Increment the index counter
                index ++;
            }
        }
    }
}
```

# Infinite Loop Cause

Infinite loops can cause great havoc in our applications. They will cause the application to fail as the Java runtime runs out of memory. Spotting the cause of an infinite loop is a critical skill to learn. Will the following code create an infinite loop? If so, can you identify the reason?

```python
  int i = 1;
  while (i != 10) {
    System.out.println("Looping and i = " + i);
    i = i + 2; # Cause of infinite loop
  }
```

- Infinite loop - caused by the fact that i will never have the value 10.
- Reason: `i = i + 2;`  will be 1,3,5...,9,11,13... So the "`i != 10`" will never be executed and hence it will continue for infinite amount of time

# For-each looping

This exercise will test your loop skills as well as require you to use your skills in method calling and int operators. You will use a for-each loop to compute the average score for a set of scores stored in an array.

class TestAverage {
    public static void main(String[] args) {
        int[] scores = {89, 82, 64, 97, 80, 79, 60, 69, 99, 100, 81, 84, 92, 96, 61, 70, 84, 91, 73, 75, 83};
        // Call the method to compute the test average with scores
        int average = computeTestAverage(scores);
        System.out.println("Test average is: " + average);
    }
    
    public static int computeTestAverage(int[] scores){
        int total = 0;
        // Loop to iterate through each of the scores
        for (int score : scores) {
            // Assign total the sum of total and the current score
            total += score;
        }
        return total/scores.length;
    }
}

# Switch vs Conditional

A switch does the same work as multiple if-then-else conditionals. In this exercise, you get to compare a conditional to a switch.

```python
class AnimalKingdom {
    public static void main(String[] args) {
        animalSpeakConditional("cow");
        animalSpeakSwitch("pig");
        animalSpeakSwitch("dog");
    }
    
    public static void animalSpeakConditional (String animal){
        if (animal.equals("cow")) {
            System.out.println("Mooo");
        } else if (animal.equals("pig")) {
            System.out.println("Oink");
        } 
        // Add another else-if conditional to check for a dog
        else if (animal.equals("dog")) {
            System.out.println("Woof");
        }
    }
    
    public static void animalSpeakSwitch (String animal) {
        switch (animal) {
            case "cow":
                System.out.println("Mooo");
                break;
            case "pig":
                System.out.println("Oink");
                break;
            // Add another case to check for a dog
            case "dog":
                System.out.println("Woof");
                // Stop the case
                break;
        }
    }
}
```

# Using a Switch

The Java Switch is used to execute a set of statements based on the comparison of an expression to a set of values. The switch is used in place of multiple if-then when the expression and values are of type byte, short, int, char or String. In this exercise, you use a switch to assign a letter grade based on a score (int) value.

```python
class AssignGrade {
    
    public static void main(String[] args) {
        assignGrade(85);
        assignGrade(92);
    }
    
    public static void assignGrade (int score){
    	char grade;
        // Add the expression from the score divided by 10
        switch ((int) score / 10) {
        	// Add the case for 10 and 9
            case 10:
            case 9:
                grade = 'A';
            // Add a statement to end the 10/9 case
                break;
            case 8:
                grade = 'B';
                break;
            case 7:
                grade = 'C';
                break;
            default:
                grade = 'F';
        }
		if (grade != 'F') {
            System.out.println("You have passed with a grade of: " + grade);
        } else {
            System.out.println("Sorry, you did not pass");
        }
    }
}
```

# Working with Data

In your final exercise of the class, you create a program to compute the average for a set of test scores. You the output the names of the students who had above average score. The student names are stored in a String array. Their corresponding test scores are stored in an int array.

This exercise tests many of your new found Java coding skills to include variable definitions, primitives, operators, arrays, looping, conditionals, and method calling.

```python
class BetterThanAverage {
    public static void main(String[] args) {
        int[] testScores = {89, 82, 64, 97, 80, 89, 70, 99};
        String[] students = {"Sally", "Muhammad", "Alex", "Ramesh", "George", "Kate", "Kelly", "Phoenix"};
        int average =  computeAverage(testScores);
        // Call on determineBest with the scores array, student names array and average score
        determineBest(students, testScores, average);
    }
    
    public static int computeAverage(int[] scores){
        int total = 0; 
        // Loop through each of the scores
        for (int score: scores){
            total += score;
        }
        // return the average score
        return total/scores.length;
    }
    
    public static void determineBest(String[] names, int[] scores, int avg){
        for (int i = 0; i < names.length; i++) {
            // Add a condition to see if a student's score is better than average
            if (scores[i] > avg) {
                System.out.println (names[i] + " scored better than average");
            }
        }
    }
}
```
