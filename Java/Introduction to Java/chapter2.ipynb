# Working with Operators

Now you start to use the data in your program, which is stored in variables, to do some interesting things. Primitive operators (arithmetic, assignment, comparison, etc.) are the means to perform operations with data. Operations used in isolation may not seem all that important, but imagine you need to write a program that starts to collect test scores and determine a person's average score and whether they passed or failed a class. This exercise helps you learn operators, but also help you see how operators are used in a wider context.

````python
class OperatorsAtWork {
   
    public static void main(String[] args) {
        int scoreTest1 = 78;
        int scoreTest2 = 94;
        // Assign testTotal the sum of scores
        int testTotal = scoreTest1 + scoreTest2;
        // Assign avg the average
        int avg = testTotal / 2;
        int passingScore = 65;
        // Incrementor passingScore
        passingScore ++;
        // Assign the isPassing variable 
        boolean isPassing = avg >= passingScore;
        System.out.println(testTotal);
        System.out.println(avg);
        System.out.println(passingScore);
        System.out.println(isPassing);
    }
}
````

# String Methods and Operators

Working with String data is a large part of any application. Java String has many methods and the concatenation operator to help you work with Strings. Getting to know them is important and in this exercise you try a few.

````python
class ShoutItOut {
    public static void main(String[] args) {
        String message = "It's a good day to have a good day";
        // Concatenate the exclamation to the message 
        String strongMessage = message + "!";
        // Concatenate the period to the message 
        String softMessage = message + ".";
        // Output the strongMessage in uppercase
        System.out.println(strongMessage.toUpperCase());
        // Output the softMessage in lowercase
        System.out.println(softMessage.toLowerCase());
    }
}
````

# Creating and Using Methods

Methods are the important building blocks by which we create our Java applications. Each method does a bit of work for the entire application. In this exercise, you will create a method to convert Fahrenheit temperatures to Celsius temperatures and call on the method with several temperatures.

````python
class TempConversion {

	// Define the temperature conversion method
    static int fahrenheitToCelsius (int fahrenheitTemperature){
    	// Compute and return the celsius temperature from the parameter
    	return (fahrenheitTemperature - 32) * 5/9;
    }

	public static void main(String[] args) {
    	// Call the method to print the conversion of 32F 
        System.out.println("32F in Celsius is: " + fahrenheitToCelsius(32));
        // Call the method to print the conversion of 90F 
        System.out.println("90F in Celsius is: " + fahrenheitToCelsius(90));
    }
}
````

# More with Methods and Operators Revisited

In this exercise, you will write a method to determine if an integer is even (versus odd). Java does have a modulo operator (%) which could be used to determine if an integer is even or odd. Another way to determine if an integer is even or odd is to divide the number by 2 and then multiply that result by 2. If the result is the same as the original number, then the number is even.

````python
class EvenOrOdd {

	// Define a method to compute and return whether a number is even
    public static boolean isEven(int aNumber){
        int result = aNumber/2;
    	// Return the result multiplied by two equals the input
        return result * 2 == aNumber;
    }
    
    public static void main(String[] args) {
    	// Call on the method with argument 5
        System.out.println("5 even: " + isEven(5));
        // Call on the method with argument 102
        System.out.println("102 even: " + isEven(102));
    }
}
````
