# Object Oriented Programming (OOP)

- OOP is a programming paradigm
- Objects = Representation to model real world entities
- Fields in objects are used to store data
- Methods in objects are used to provide behavior
- Classes are blueprints for creating objects
- Objects are instances of a class
- Other than primitive data types, everything else belong to objects

```python
class Dog {
    // fields
    String name;
    int age;
    String bark;
    // constructor to map fields during object initialization
    public Dog(String name, int age) {
        this.name = name;
        this.age = age;
        this.bark = "Woof!";
    }

    void bark() { // method
        System.out.println(bark + " name = " + name + ", age = " + age);
    }
}

public class DogDaycare {
    Dog polo;
    Dog aki;
    public DogDaycare() {
        // `new` keyword to create objects from class "Dog"
        this.polo = new Dog("Polo", 4);
        this.aki = new Dog("Aki", 3);
    }
    void makeDogsBark() {
        polo.bark(); // Output: Woof! name = Polo, age = 4
        aki.bark(); // Output: Woof! name = Aki, age = 3
    }
}
```
