# Object Oriented Programming (OOP)

- OOP is a programming paradigm
- Objects = Representation to model real world entities
- Fields in objects are used to store data
- Methods in objects are used to provide behavior
- Classes are blueprints for creating objects
- Objects are instances of a class
- Other than primitive data types, everything else belong to objects
- Access modifiers: Defines how members of a class can be accessed outside of the class. 
    - private, 
    - public,
    - protected,
    - default
- Encapsulation: We do not want outside code to change internal state of objects directly. So we use `private` fields to ensure encapsulation
- Inheritance : Allows us to create hierarchies in classes and share common attributes (fields) and behaviors (methods) of classes
    - parent class : super class
    - child class : sub class
    - child class inherits code from parent class
    - child class constructor **MUST** use `super()` to call parent class constructor before anythin else
- Abstract class : Does not allow object creation, but allows to extend itself.
    - used to create concept of top level hierarchy instead of model itself when the top level class is too generic to create any object. So, we use it to restrict any direct object creation from this generic class using `abstract`
- Interfaces: Establish a set of rules (abstract methods without body) so that any class accept the rules (implement the interface) must be obedient in executing all the rules (implement the methods by overriding them and creating required body)
    - Interface as data-type : Allows polymorphism by using a variable of it's type to store an object of a class that implements it
- Polymorphism: Multiple forms
    - Overriding or runtime polymorphism: 
        - changing behavior of a class inside it's subclass. 
        - method signature does not change. Only method implementation change
    - Overloading:
        - changing behavior of a class inside it or it's subclass. 
        - method signature change (same method name, but contains different parameter list)

<center><img src="images/02.01.png"  style="width: 400px, height: 300px;"/></center>


```python
class Dog {
    // fields: making them private to ensure encapsulation
    private String name;
    private int age;
    private String bark;
    // constructor to map fields during object initialization
    public Dog(String name, int age) {
        this.name = name;
        this.age = age;
        this.bark = "Woof!";
    }

    public void bark() { // method, allowed behavior outside class, so public modifier is used
        System.out.println(bark + " name = " + name + ", age = " + age);
    }
}

public class DogDaycare {
    Dog polo;
    Dog aki;
    public DogDaycare() {
        // `new` keyword to create objects from class "Dog"
        this.polo = new Dog("Polo", 4);
        this.aki = new Dog("Aki", 3);
    }
    public void makeDogsBark() {
        polo.bark(); // Output: Woof! name = Polo, age = 4
        aki.bark(); // Output: Woof! name = Aki, age = 3
    }
}
```

# Inheritance

<center><img src="images/02.03.png"  style="width: 400px, height: 300px;"/></center>


```python
public class Animal {
    protected name;
    public Animal(String name) { // Superclass constructor
        this.name = name;
    }
}

public class Dog extends Animal {
    private String breed;
    public Dog(String name, String breed) { // Subclass constructor
        super(name); // Superclass constructor call
        this.breed = breed;
        System.out.println(" name = " + super.name + ", breed = " + breed);
    }
}
```

# Abstract Class

<center><img src="images/02.02.png"  style="width: 400px, height: 300px;"/></center>


```python
public abstract class Animal { // Abstract class, restricting any object creation from this TOO GENERIC class
    protected String name;
    protected int age;
    protected String animalClass;
}

public class Dog extends Animal { // Extending Abstract class to inherit all the codes from Animal class
    private int height;
}

public class Snake extends Animal { // Extending Abstract class to inherit all the codes from Animal class
    private boolean venomous;
}
```

# Interfaces

```python
public interface Animal { // Interface definition
    String getSound(); // Method signature
}

public class Dog implements Animal { // Implementing the interface
    @Override // Override the interface method
    public int getSound() {
        return "Woof!";
    }
}

public class Snake implements Animal { // Implementing the interface
    @Override // Override the interface method
    public int getSound() {
        return "Hiss!";
    }
}
public class Main {
    public static void main(String[] args) {
        Animal a1 = new Dog(); // LOOK ON THE LEFT: Subtyping to interface to ensure polymorphism
        System.out.println(a1.getSound()); // Woof!
    }
}
```

# Polymorphism

### Overriding

```python
class Animal {
   public void displayInfo() {
      System.out.println("I am an animal.");
   }
}

class Dog extends Animal {
   @Override
   public void displayInfo() {
      System.out.println("I am a dog.");
   }
}

class Main {
   public static void main(String[] args) {
      Dog d1 = new Dog();
      d1.displayInfo();
   }
}
```

### Overloading

```python
class Calculate
{
  void sum (int a, int b)
  {
    System.out.println("sum is"+(a+b)) ;
  }
  void sum (float a, float b)
  {
    System.out.println("sum is"+(a+b));
  }
  Public static void main (String[] args)
  {
    Calculate  cal = new Calculate();
    cal.sum (8,5);      //sum(int a, int b) is method is called.
    cal.sum (4.6f, 3.8f); //sum(float a, float b) is called.
  }
}
```
