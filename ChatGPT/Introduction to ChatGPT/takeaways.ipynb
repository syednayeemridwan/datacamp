# Chapter 1

- Generates text using LLM
- Generative AI: Generates new content
- ChatGPT is great at:
    - summarizing reports
    - interpret compplex information
- Limitations:
    - knowledge cutoff = no knowledge beyond a certain date for specific version of chatgpt
    - training bias = data that used for training LLM could contain biases. so, the response could also be biased.
    - context tracking = can struggle to keep track of context if focus shifts. can lead to inaccurate or irrelevant result.
        - Solution: keep conversations to a single topic
    - Hallucination = provides inaccurate information confidently. occurs when it goes beyond the model's knowledge or capabilities.
- Prompt Engineering: process of writing prompt to maximize the quality and relevance of the response. Use keywords for more presonalizations. 
- Great prompt guidelines:
    - Be clear and specific on necessary information
    - Do not give extra unnecessary information that is out of context
    - Use correct grammar and spelling
    - Provide examples
- Challenge:
    - Quantity of data makes detecting bias prior to training difficult
        - Goal: Develop more robust bias mitigation procedures
- Large Language Models (LLMs)
    - Learns from a huge text dataset
    - Algorithms detect patterns in text
    - Fine-tune the model by rating responses
