# Chapter 1

- Early LLM
    - predicts next word in a sequence : James likes to eat (pizza)
    - Predicts Masked word in the middle:  Sasha (loves) football
- RNN creates the dawn of LLM:
    - process data sequentially
    - remembers the words in sequence
    - Uses hidden states to retain memory
    - struggles to maintain relevance and context when sequence is long (due to vanishing gradient problems)
    - Rescue: Groudbreaking architecture of the transformer (weighs the significance of words irrespective of position, process inputs simultaneously, attention mechanisms)
- Tokenization:
    - transforms words into something that computers can understand 
    - holds information about syntax (sentence structure) and symantics (meaning)
    - numerical representation of pieces of text
- Vectorization:
    - Comes after tokenization
    - numerical arrays that capture the symantic meaning of the tokens
    - This process is called embedding
    - At the end, decoding translates the generated sequence of token ids back to words and sentences
- Transformer:
    - Unlike RNN, it does not lose information for long texts
    - Self attention: looks at inputs and outputs to learn without using sequence-aligned RNNs
    - Allows to weigh importance of each token in the input relative to all other tokens
    - Capture relationship and dependencies between tokens regardless the distance between words
    - Quantifies the relationship between tokens into attention score for each tokens importance
    - uses the attention score to focus most relevant / important tokens and build contextual understanding.
