In this exercise, you're going to practice exploring dictionaries and retrieving data stored in a simple dictionary. The dictionary avocados is loaded into your workspace, which includes information about a single purchase of avocados.

# Modifying dictionaries


Like lists, dictionaries can be changed. You can add new key:value pairs, update existing pairs, remove pairs. In this exercise, you'll explore creating and modifying dictionaries


```python
# Create dict with milk and apples
groceries = {'gallons of milk': 1,
             'apples': 3}
print(groceries)

# Add candy bars to groceries dict
groceries['candy bars'] = 2
print(groceries)

# Add apple to groceries
groceries['apples'] = 5
print(groceries)

# Remove milk from groceries
del groceries['gallons of milk']

print(groceries)
```

# Plot data from a DataFrame


pandas DataFrames are one of the most frequently used data structures in Python. They consist of clearly organized and labeled data, which make it easy to start exploring your data quickly. One of the best ways to explore a DataFrame is by employing the power of matplotlib. In this exercise, you'll perform a visualize exploration of bicycle crossings over the Freemont Bridge in Seattle held in the freemont_df DataFrame.


```python
import matplotlib.pyplot as plt

freemont_df = pd.read_csv("dataset/fremont_bridge_average_daily_by_month.csv")

# Create a bar plot of mean daily trips
plt.barh(freemont_df.columns, 
         freemont_df.mean(), 
         xerr=freemont_df.sem(), 
         capsize=3)

# Add an x-label and plot title
plt.xlabel('Average Trips')
plt.title('Average trips per day')

# Remember to show the plot
plt.show()
```

```python
# Create a line plot with a legend
plt.plot(freemont_df)
plt.legend(freemont_df.columns,
           loc='best',
           frameon=False)

# Add labels and plots title
plt.xlabel('Hour')
plt.ylabel('Trips')
plt.title('Hourly trips by day')

# Remember to show the plot
plt.show()
```

# The many flavors of .iloc


There are multiple ways to index a pandas DataFrame – maybe too many. In this exercise, you're going to explore the many usages – or pitfalls of the .iloc indexer. This is something that will likely trip you up on your Python adventure, so it's good to know up front what you'll be up against.


```python
# Access first row and first column
iloc_1 = animals.iloc[0,0]
print(iloc_1)
print(type(iloc_1))

# Access first row
iloc_2 = animals.iloc[0]
print(iloc_2)
print(type(iloc_2))

# Access first row with double brackets
iloc_3 = animals.iloc[[0]]
print(iloc_3)
print(type(iloc_3))
```
