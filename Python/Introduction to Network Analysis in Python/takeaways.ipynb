```
import networkx as nx
G = nx.Graph() # Initialize empty graph
G.add_nodes_from([1, 2, 3]) # Create 3 nodes in the graph
G.nodes() # See all nodes
len(G.nodes()) # See no of nodes nodes
G.nodes(data=True) # See all nodes with data
list(G.neighbors(5)) # Find all nodes that are connected with node 5
nx.degree_centrality(G) # actual neighbours of a node / total possible neighbours for that node, dict
list(nx.degree_centrality(G).values()) # find values of degree of centrality, use histogram to get idea
nx.betweenness_centrality(G) # total shortest paths through a given node / all possible shortest paths, dict
list(nx.betweenness_centrality(G).values()) # find values of betweenness of centrality
G.add_edge(1, 10) # Create edge between node 1 and 10
G.edges[1,10]['weight'] = 2 # provide info / weight for edge between  1 and 10, can be considered a dictionary
G.edges() # See all edges
list(G.edges(data=True)) # See all edges in list along with data
G.nodes[1]['label'] = 'blue' # provide information for node 1, each node can be considered a dictionary
nx.draw(G)
import matplotlib.pyplot as plt
plt.show()
G = nx.Graph() # Non-directional graph
D = nx.DiGraph() # Directional graph
M = nx.MultiGraph() # Multi non-Directional graph
MD = nx.MultiDiGraph() # Multi-Directional graph
# Custom package for visualization
import nxviz as nv # Find documentation for API at : https://ericmjl.github.io/nxviz/
import matplotlib.pyplot as plt
ap1 = nv.arc(G) # Visualize with arcplot, where nodes are ordered along axis and arc line defines relationship
ap2 = nv.matrix(G) # Visualize graph with matrixplot to see if an edge exists between pairs of nodes
ap3 = nv.circos(G) # Visualize graph with cicos plot, forms circle, used in genomics
plt.show()

# Convert T to a matrix format: A
A = nx.to_numpy_matrix(T)
# Convert A back to the NetworkX form as a directed graph: T_conv
T_conv = nx.from_numpy_matrix(A, create_using=nx.DiGraph())
```
