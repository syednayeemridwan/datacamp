# Univariate plots in pandas


You'll start this chapter by using the plotting methods in pandas. This is typically done by calling the .plot() method on the relevant column, and passing in a value for the kind argument.


```python
import pandas as pd
tips = pd.read_csv("dataset/tips.csv")
tips.head()
```

```python
import matplotlib.pyplot as plt

# Histogram of tip
tips.tip.plot(kind = 'hist')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Boxplot of the tip column
tips.tip.plot(kind = 'box')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Bar plot
cts = tips.sex.value_counts()
cts.plot(kind='bar')
plt.show()
```

# Bivariate plots in pandas


Comparing multiple variables simultaneously is also another useful way to understand your data. When you have two continuous variables, a scatter plot is usually used.


```python
import matplotlib.pyplot as plt

# Scatter plot between the tip and total_bill
tips.plot(x='total_bill', y='tip', kind='scatter')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Boxplot of the tip column by sex
tips.boxplot(column='tip', by='sex')
plt.show()
```

# Univariate plots in seaborn


Seaborn is a popular plotting library. It embraces the concepts of "tidy data" and allows for quick ways to plot multiple variables.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Bar plot
sns.countplot(x='sex', data=tips)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Histogram
sns.histplot('total_bill', kde = True)
plt.show()
```

# Bivariate plots in seaborn


Boxplots can be used on univariate or bivariate data. We use boxplots when we have a numeric variable and a categorical variable. Scatter plots are used when we have two numeric variables.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot for tip by sex
sns.boxplot(y='tip', x='sex', data=tips)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plot of total_bill and tip
sns.regplot(x='total_bill', y='tip', data=tips)
plt.show()
```

# Facet plots in seaborn


Some plotting functions in seaborn such as histplot() and lmplot() have built-in facets. All you need to do is pass a col and/or row argument to create facets in your plot.

For functions that do not have built-in facets, you can manually create them with the FacetGrid() function, and then specify the col and/or row to create your facets.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plot of total_bill and tip faceted by smoker and colored by sex
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex', col='smoker')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# FacetGrid of time and smoker colored by sex
facet = sns.FacetGrid(tips, col="time", row="smoker", hue="sex")

# Map the scatter plot of total_bill and tip to the FacetGrid
facet.map(plt.scatter, 'total_bill', 'tip')
plt.show()
```

# Univariate and bivariate plots in matplotlib


Matplotlib is at the core of all the plotting functions in Pandas and Seaborn. Understanding how matplotlib works will help you tweak your plot for publication or reports.

# Subfigures in matplotlib


Sometimes you want to combine several subplots in a single figure.

You can do this by creating a figure and an axes simultaneously by using the plot.subplots() function.


```python
import matplotlib.pyplot as plt

# Create a figure with 1 axes
fig, ax = plt.subplots(1, 1)

# Plot a scatter plot in the axes
ax.scatter(tips.tip, tips.total_bill)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a figure with scatter plot and histogram
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.scatter(tips['tip'], tips['total_bill'])
ax2.hist(tips['total_bill'])
plt.show()
```

# Working with axes


Now that you've seen how to create a figure within matplotlib, you can extend this knowledge to the other plotting libraries you've used. If you look at what gets returned from the various seaborn plotting calls, most functions will return an axes, while some will return a figure. You can leverage this information when creating axes' in a figure. This way you are not limited to just using base matplotlib as a plotting library when making more complex figures.


```python
# histplot of tip
import seaborn as sns
dis = sns.histplot(tips['tip'], kde = True)

# Print the type
print(type(dis))
```

```python
# Figure with 2 axes: regplot and histplot
fig, (ax1, ax2) = plt.subplots(1,2)
sns.histplot(tips['tip'], ax=ax1, kde = True)
sns.regplot(x='total_bill', y='tip', data=tips, ax=ax2)
plt.show()
```

# Polishing up a figure


Since plt.subplots() returns axes object, you can use the .set_title(), .set_xlabel(), and .set_ylabel() methods to add labels to your figure.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure with 1 axes
fig, ax = plt.subplots()

# Draw a histplot
ax = sns.histplot(tips['total_bill'], kde = True)

# Label the title and x axis
ax.set_title('Histogram')
ax.set_xlabel('Total Bill')
plt.show()
```
