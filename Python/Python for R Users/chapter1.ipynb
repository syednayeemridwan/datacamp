# Assignment and data types


This exercise covers assigning values to variables and the basic data types in Python


```python
# Assign an integer (42) to w
w = 42

# Assign a float (3.14) to x
x = 3.14

# Assign a boolean (True) to y
y = True

# Assign a string ('python') to z
z = 'python'

# Print the data types of y and z
print(type(y))
print(type(z))
```

# Lists


Lists are generic containers that can store heterogeneous data in Python. In this example, you'll store data about a person in a list. The list will contain the following data (in that order):

- First name
- Last name
- Sex
- Employment status
- Number of twitter followers


```python
# Assign the values to the list
person_list = ['Hugo', 'Bowne-Anderson', 'male', True, 3509]

# Get the first name from the list
print(person_list[0])

# Get the first and last name from the list
print(person_list[0:2])

# Get the employment status
print(person_list[-2])
```

# Dictionaries


Dictionaries are similar to named vectors and lists in R in that they provide a name (or key) for you to access values. In Python, dictionaries are created using a pair of curly brackets, { }, and passed in key-value pairs. Keys and values are separated with a colon.


```python
# Create a dictionary form of the employee information list
person_dict = {
    'fname': person_list[0],
    'lname': person_list[1],
    'sex': person_list[2],
    'employed': person_list[3],
    'twitter_followers': person_list[4]
}

# Get the first and last names from the dict
print(person_dict['fname'])
print(person_dict['lname'])
```

# Methods


There will be times when you need to add values into a list, or update values in a dictionary. Here we will see some of the object-oriented nature of Python, which is different from R.


```python
# Append values to a list
person_list.append(2018)

# Print person_list
print(person_list)

# Print the last element of person_list
print(person_list[-1])

# Update the person_dict dictionary
person_dict.update({'date':'2018-06','twitter_followers':3571})

# Print the person_dict dictionary
print(person_dict)
```
