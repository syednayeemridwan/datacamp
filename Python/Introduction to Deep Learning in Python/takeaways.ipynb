### Neural network

- Representation learning : The subsequent layers build increasingly sophisticated representations of raw data (first layers find simple patterns like lines, later layers combine those simple patterns to find more complex geometric shapes)
- partially replace the need for feature engineering
- In linear regression or other simple models, we assume the independent variables have minimal or no interaction with each other.
- In deep learning, we take for granted that there are always some level of interaction among independent variables.
    - The nodes in the hidden layers are the aggregation of information of independent variables that adds the models ability to capture the interaction
    - The more nodes we have, the more interaction we capture
- Forward propagation:
    - Used to calculate values of hidden layers' nodes
    - Creates a nework of nodes that process data by passing the data from input to output layer for prediction
    - All nodes from previous layer is connected to each node of the next layer with a line
    - The line has a weight that indicates how the previous nodes affect the next layer's hidden node
    - The value of the hidden node is the added multiplication of weights and previous node's values
    - Activation function is used for transformation of calculated value
- Activation function:
    - Allows the model to capture non-linear pattern (Anything that is not a straight line)
    - It takes the value of the hidden layer's node and transforms that value into a meaningful value. (eg: tanh, relu)
- Loss function:
    - Aggregates errors in predictions from many data points into single score
    - Measure of model's predictive performance (lower loss = better model)
    - Goal : Find the weights that give the lowest value for the loss function (gradient descent)
    - We use slope to find the minimum loss. (As long as slope is not close to zero, keep going downhill or opposite to slope)
    - learning rate : how much do we update the weight after each step (Update each weight by subtracting learning rate * slope from current value)
- Back propagation:
    - Change in adjusted weights (value of edges) improve model performance from initial random assigned weights

```
import numpy as np

def relu(input):
    '''Define your relu activation function here'''
    # Calculate the value for the output of the relu function: output
    output = max(input, 0)
    
    # Return the value just calculated
    return(output)

def tanh(input):
    '''Define your hyperbolic tangent activation function here'''
    # Calculate the value for the output of the tanh function: output
    output = np.tanh(input)
    
    # Return the value just calculated
    return(output)

input_data = np.array([3, 5])
weights = {'node_0_0': np.array([2, 4]),
 'node_0_1': np.array([ 4, -5]),
 'node_1_0': np.array([-1,  2]),
 'node_1_1': np.array([1, 2]),
 'output': np.array([2, 7])}

  
def predict_with_network(input_data):
    # Calculate node 0 in the first hidden layer
    node_0_0_input = (input_data * weights['node_0_0']).sum()
    node_0_0_output = relu(node_0_0_input)

    # Calculate node 1 in the first hidden layer
    node_0_1_input = (input_data * weights['node_0_1']).sum()
    node_0_1_output = relu(node_0_1_input)

    # Put node values into array: hidden_0_outputs
    hidden_0_outputs = np.array([node_0_0_output, node_0_1_output])
    
    # Calculate node 0 in the second hidden layer
    node_1_0_input = (hidden_0_outputs * weights['node_1_0']).sum()
    node_1_0_output = relu(node_1_0_input)

    # Calculate node 1 in the second hidden layer
    node_1_1_input = (hidden_0_outputs * weights['node_1_1']).sum()
    node_1_1_output = relu(node_1_1_input)

    # Put node values into array: hidden_1_outputs
    hidden_1_outputs = np.array([node_1_0_output, node_1_1_output])

    # Calculate model output: model_output
    model_output = (hidden_1_outputs * weights['output']).sum()
    
    # Return model_output
    return(model_output)

output = predict_with_network(input_data)
print(output)
```

# Backward Propagation

- slope of output node for the last hidden layer nodes = 2 * (actual value - predicted value in output node, that you get using forward propagation) * incoming input node value * slope of activation function, this is 1 for relu
- slope of weight = input node value * slope of output node * activation function slope (1 for relu)
- updated weight = weight or edge - learning rate * slope of weight
- backpropagated value in nodes other than inputs = last layer node * error * previous layers node value * slope of activation function
- It is common to calculate slopes on only a subset of the data (a batch) for computational efficiency
- Use a different batch of data to calculate the next update
- Start over from the beginning once all data is used
- Each time through the training data is called an epoch
- When slopes are calculated on one batch at a time: stochastic gradient descent
- backpropagation takes the prediction error from output layer and propagates it to the input layers through the hidden layers.
- Thus, it allows gradient descent to update all weights in the neural network (chain rule of calculus)
- Slope of node values are the sum of the slopes for all weights that come out of them
- Previous node value = node value * slope

```
weights = np.array([0, 2, 1])
input_data = np.array([1, 2, 3])
target = 0
# Set the learning rate: learning_rate
learning_rate = 0.01

# Calculate the predictions: preds (forward propagation)
preds = (input_data * weights).sum()

# Calculate the error: error
error = preds - target

# Calculate the slope of each input: slope
slope = 2 * error * input_data

# Update the weights: weights_updated (Backpropagation)
weights_updated = weights - learning_rate*slope

# Get updated predictions: preds_updated (forward propagation)
preds_updated = (weights_updated * input_data).sum()

# Calculate updated error: error_updated
error_updated = preds_updated - target

# Print the original error
print(error)

# Print the updated error
print(error_updated)
```

```
weights = np.array([0, 2, 1])
input_data = np.array([1, 2, 3])
target = 0

def get_slope(input_data, target, weights):
    error = get_error(input_data, target, weights)
    slope = 2 * input_data * error
    return(slope)

def get_error(input_data, target, weights):
    preds = (weights * input_data).sum()
    error = preds - target
    return(error)

def get_mse(input_data, target, weights):
    errors = get_error(input_data, target, weights)
    mse = np.mean(errors**2)
    return(mse)

import matplotlib.pyplot as plt
n_updates = 20 # epochs
learning_rate = 0.01
mse_hist = []

# Iterate over the number of updates
for i in range(n_updates):
    # Calculate the slope: slope
    slope = get_slope(input_data, target, weights)
    
    # Update the weights: weights
    weights = weights - learning_rate * slope
    
    # Calculate mse with new weights: mse
    mse = get_mse(input_data, target, weights)
    
    # Append the mse to mse_hist
    mse_hist.append(mse)

# Plot the mse history
plt.plot(mse_hist)
plt.xlabel('Iterations')
plt.ylabel('Mean Squared Error')
plt.show()

```

# Dying Neurons and Vanishing Gradients

- Duying Neuron Problem:
	- This happens when a neuron takes a value less than 0 for all the rows of the data
    - When relu is used and when the slope is 0, 
    	updated weight = current weight - node val * slope
        => updated weight = current weight (no change,  neuron comes to a standstill)
	- Occurs when a neuron keeps getting negative inputs
    - Solution: Use different activation function other than relu (eg: tanh). But tanh will produce vanishing gradient problem
- Vanishing Gradients:
	- Prevents neural networks from booming sooner (when updates to weights 
		during backpropagation are close to 0)
	- Happens when sigmoid(tanh) activation function is used
	- Occurs when many layers of neural network have very small slopes (e.g. due to
		being on the flat part of the tangent curve
	- Sigmoid function makes the intermediate values of the neural network 
		between 0 and 1 
	- When back-propagation is used, we keep multiplying values less 
		than 1 with each other. 
	- So the gradient keeps getting smaller and smaller moving 
		backward to the network.
	- So, the neurons in the earlier layers learn slowly 
		compared to the neurons in the later layers in the network
	- Thus, training takes too long and accuracy is compromised.
