### Common Functions


- `max()`
- `min()`
- `sum()`
- `round()`
- `len()`
- `sorted()`
- `sum()`
- `help()`
- `print()`
- Functions vs methods:
  - Function = code to perform a task (eg : `sum(some_list)`)
  - Method = a function that is specific to a data type (eg : `df.head()`)


### Common Modules


- Module : Python file
- packages : Collection of modules
- `os` : For interacting with OS
- `collections` : Advanced data structure types and functions
- `string` : String operations
- `logging` : Log information when testing or running software
- `subprocess` : Run terminal commands
- More on : https://docs.python.org/3/py-modindex.html
- See doc : `help(module_name)`
- Install package : `pip install package_name`
- Use module of a package : `from package_name import module_name`
