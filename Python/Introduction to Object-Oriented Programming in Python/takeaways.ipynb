### What is `self` in python?

```
class Customer:
    def identify(self, name):
        print("I am Customer " + name)
cust = Customer()
cust.identify("Laura")
```

here, `self` acts as a hook to tag the method `identify` with any object created from `Customer` class.

for example, `cust.identify("Laura")` will be interpreted as `Customer.identify(cust, "Laura")`

we can do the same for attributes. Our improvement of the above code can be done by using a setter for tagging an attribute:
```
class Customer:
    def set_name(self, new_name): # Setter method
        self.name = new_name # Creating an attribute
    def identify(self):
        print("I am Customer" + self.name) # Calling the attribute value

cust = Customer()
cust.set_name("Rashid Volkov") # setting an attribute
cust.identify() # Accessing the attribute value
```

here, `self` acts as a hook to tag the `name` variable with any object created from `Customer` class.

for example, `self.name` will be interpreted as `Customer.name`

### What is `__init__` in python?


```
class Customer:
    def __init__(self, name, balance=0): # Set a default value for balance
        self.name = name
        self.balance = balance # Assign the new attribute
        print("The __init__ method was called")

# Don't specify the balance explicitly
cust = Customer("Lara de Silva")
print(cust.name) # "Lara de Silva"
# The balance attribute is created anyway
print(cust.balance) # 0
```
