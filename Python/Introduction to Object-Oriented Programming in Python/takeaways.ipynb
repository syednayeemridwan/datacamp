### What is `self` in python?

```
class Customer:
    def identify(self, name):
        print("I am Customer " + name)
cust = Customer()
cust.identify("Laura")
```

here, `self` acts as a hook to tag the method `identify` with any object created from `Customer` class.

for example, `cust.identify("Laura")` will be interpreted as `Customer.identify(cust, "Laura")`

we can do the same for attributes. Our improvement of the above code can be done by using a setter for tagging an attribute:
```
class Customer:
    def set_name(self, new_name): # Setter method
        self.name = new_name # Creating an attribute
    def identify(self):
        print("I am Customer" + self.name) # Calling the attribute value

cust = Customer()
cust.set_name("Rashid Volkov") # setting an attribute
cust.identify() # Accessing the attribute value
```

here, `self` acts as a hook to tag the `name` variable with any object created from `Customer` class.

for example, `self.name` will be interpreted as `Customer.name`

### What is `__init__` in python?


```
class Customer:
    def __init__(self, name, balance=0): # Set a default value for balance
        self.name = name
        self.balance = balance # Assign the new attribute
        print("The __init__ method was called")

# Don't specify the balance explicitly
cust = Customer("Lara de Silva")
print(cust.name) # "Lara de Silva"
# The balance attribute is created anyway
print(cust.balance) # 0
```

### Class-level attributes and methods

```
class Employee:
    MIN_SALARY = 30000 # class attribute, applied for global constraint, not instance attribute
    def __init__(self, name, salary): self.name = name
        # Use class name
        # to access class attribute
        if salary >= Employee.MIN_SALARY: 
            self.salary = salary
        else: 
            self.salary = Employee.MIN_SALARY # Accessing class attribute inside class (no self keyword)

emp1 = Employee("John",40000)
print(emp1.MIN_SALARY) # Accessing class attribute using object
```
- quite like the `static` variable. use class name to access these attributes
- Class attributes CAN be inherited, and the value of class attributes CAN be overwritten in the child class

```
class Employee:
    def __init__(self, name, salary): 
        self.name = name 
        self.salary = salary
    @classmethod
    def from_file(cls, filename): # This can be used as second constructor
        with open (filename,"r") as f: 
            name = f.readline().strip() 
            salary = int(f.readline().strip())
        return cls(name, salary) # this will call the __init__ method

# Create an employee without calling Employee()
emp = Employee.from_file("employee_data.txt") # AAccessing class level method
print(emp.name)
```
- quite like the `static` method. use `clas` along with `@classmethod` to define these methods
- there can only be one `__init__` method in each class, so if there is any need to create multiple constructor, this is the way
- This type of method ensures singleton property (restrict class to a single object) and `return cls(name, salary)` actually calls `__init__` method
- These methods don't require instance-level attributes


### Inheritance

```
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def withdraw(self, amount):
        self.balance -= amount

# Class inheriting from BankAccount
class SavingsAccount(BankAccount):
    def __init__(self, balance, interest_rate):
        BankAccount.__init__(self, balance) # Call ParentClass constructor (optional)
        self.interest_rate = interest_rate # new attribute for this class
    def compute_interest(self, n_periods=1): # New method for this class
        return self.balance * ( (1 + self.interest_rate) ** n_periods - 1)

class CheckingAccount(BankAccount):
    def __init__(self, balance, limit):
        BankAccount.__init__(self, balance) # Call ParentClass constructor (optional)
        self.limit = limit  # new attribute for this class
    def deposit(self, amount): # New method for this class
        self.balance += amount
    def withdraw(self, amount, fee=0): # OVERRIDING method for this class
        if amount <= self.limit:
            BankAccount.withdraw(self, amount + fee)
        else:
            pass # Won't run if the condition isn't met

savings_acct = SavingsAccount(1000)
type(savings_acct) # Check data/object type
savings_acct.balance # Attribute inherited from BankAccount
savings_acct.withdraw(300) # Method inherited from BankAccount
savings_acct = SavingsAccount(1000)
# Checking if an object is instance of both child class and parent class (is-a relationship)
isinstance(savings_acct, SavingsAccount) # True
isinstance(savings_acct, BankAccount) # True

check_acct = CheckingAccount(1000, 25)
check_acct.withdraw(200) # Will call withdraw from CheckingAccount
check_acct.withdraw(200, fee=15) # Will call withdraw method from CheckingAccount
bank_acct = BankAccount(1000)
bank_acct.withdraw(200) # Will call withdraw method from BankAccount
bank_acct.withdraw(200, fee=15) # Will produce error, Violates polymorphism : Parent / child classes have different method signature
```
