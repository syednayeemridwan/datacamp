# Data

- Structured : Organized, tabular, typically text and numbers
- Unstructured : Unorganized, difficult to manage, no format, typically audio video and texts
- Quantitative content : Numerical data (expressed with numbers like height or temperature)
- Qualitative content : Categorical data (expressed with group like language, country etc)
- Meta Data : Information about data. Provides meaning about context.
    - eg: Timeframe, location etc
- Goal: support business objectives
    - Profitability
    - Social good
    - Research
    - Customer satisfaction & employee
    - happiness
- How: by improving decision making
    - Measure return on investment (ROI)
    - Optimize processes and find new opportunities
- Examples:
    - supply chain management : The sequence of processes involved in the production and distribution of a product
        - Demand forecasting
    - Data in healthcare : wearable devices to gather personal health data
        - Detect and prevent health problems

# DIKW Pyramid (Data Information Knowledge Wisdom)

- Data : Stand-alone pieces of information
- Information : We need to add meaning or context to raw data / data points (eg: Column names, index values)
- Knowledge : Connecting the dots and pieces of information together
- Wisdom : Uncovering insights and hidden patterns (Apply findings to guide actions for better decisions)

# Data Driven Decision Making Process

1. A good question : Ask exact question
2. Gather Data : Identify the sources and gather them
3. Prepare Data : Clean bad data in good format (80% of time is spent here)
4. Decisions based on Analysis : Find insights based on available information
5. Make decisions: Balance outcome with knowledge (+personal experience) that drives the most optimized outcome

# Data Utilization as Resource

- Aggregate/summarize to avoid being overwhelmed / lose focus 
- Target Data measures based on the business requirement : Metrics, KPIs, Benchmarks
- Being curious : Trace for validations of assumptions and investigate
