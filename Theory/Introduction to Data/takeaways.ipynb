# Data

- Structured : Organized, tabular, typically text and numbers
- Unstructured : Unorganized, difficult to manage, no format, typically audio video and texts
- Quantitative content : Numerical data (expressed with numbers like height or temperature)
- Qualitative content : Categorical data (expressed with group like language, country etc)
- Meta Data : Information about data. Provides meaning about context.
    - eg: Timeframe, location etc
- Goal: support business objectives
    - Profitability
    - Social good
    - Research
    - Customer satisfaction & employee
    - happiness
- How: by improving decision making
    - Measure return on investment (ROI)
    - Optimize processes and find new opportunities
- Examples:
    - supply chain management : The sequence of processes involved in the production and distribution of a product
        - Demand forecasting
    - Data in healthcare : wearable devices to gather personal health data
        - Detect and prevent health problems

# DIKW Pyramid (Data Information Knowledge Wisdom)

- Data : Stand-alone pieces of information
- Information : We need to add meaning or context to raw data / data points (eg: Column names, index values)
- Knowledge : Connecting the dots and pieces of information together
- Wisdom : Uncovering insights and hidden patterns (Apply findings to guide actions for better decisions)

- Aggregate/summarize to avoid being overwhelmed / lose focus 
- Target Data measures based on the business requirement : Metrics, KPIs, Benchmarks
- Being curious : Trace for validations of assumptions and investigate


# Data Flow

- Pipeline of data from source to consumption
- Domains:
    - Data Governance : Ensures consistency, trustworthiness, not misused
    - Data Quality : Data is accurate, valid, complete
    - Data Privacy and Security : Data is protected, accessed and used according to regulations

# Data Ethics

- Ensures that data is collected, stored, and used ethically
- Five principles
    1. Permission : Consent before taking the data
    2. Transparency : How the data will be collected, stored and used
    3. Privacy : PII or Sensitive information that can be traced to a person should be protected
    4. Intentions : Data is used for the right reasons
    5. Outcome: The outcome should not harm anyone

# Data Lifecycle

- Ensure data is regulated responsibly, increase trustworthiness of a company
- Identify potential areas for improvement
- Improve efficiency and effectiveness of operations

- Planning and collecting: Business question and the type of data and how it will be managed, how to achieve best result
- Storing and managing: Data should be stored so that it can be easily accessible
- Cleaning and processing: Deal with messy data (missing data / inconsistent types)

# Common Mistakes

- Not having a clear goal or question : Should not be a vague question
- Insufficient or wrong data: Collect data that answers the research question and not pose data bias
- Lack of appropriate analysis: Do not jump to conclusions too early and incorrect aggregations
- No clear communication of results: Should not be too technical or avoid cherry-picking the result
