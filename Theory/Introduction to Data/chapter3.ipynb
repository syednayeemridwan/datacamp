# Ethics and privacy

Ethics and privacy are interconnected terms. A breach of privacy will not only cost your company financially, but it will also lead to the loss of customers and damage your reputation as a result. It is both an individual and governmental responsibility to protect our data.

# Ordering the data life cycle

Data goes through various phases in the data life cycle. These phases are:

- Planning and collecting
- Storing and managing
- Cleaning and processing
- Analyzing and visualizing
- Sharing
- Archiving/destroying
