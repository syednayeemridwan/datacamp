# Paradigm

- A way of thinking about a topic
- eg: First we thought the earch was flat, then we knew that the earth is round (This is example of paradigm shift)
- Programming Paradigm : A way of approaching writing a program
- Same programming paradigm may/can be used by many languages
- Same programming language can/may use by many paradigms
- python can be used for all procedural, functional and object oriented programming paradigms
- Different programming paradigms requires different approach to solve in optimal time and effort
- Separation of responsibilities are different for different paradigms

# Imperative vs Declarative Programming paradigms

- Imperative:
    - How the program should be executed
    - Requires step-by-step instructions
    - eg: Flowchart
    - Procedural and OOP programming
    - More common
- Declarative:
    - What the program should execute
    - Programmer defines logic instead of exact steps
    - eg: sudoku (rules are specified, but exact solution is not known)
    - Functional programming
    - Less common, not for general purpose

# Modular Code


- Code is broken into re-usable sections
- Avoid repeatition
- Separation of responsibilities
- Reduce chance of introducing bugs
- Saves development time
- Easier to understand

```python
# Non- Modular code
sum_values1 = 1 + 2
avg_values1 = sum_values1 / 2
sum_values2 = 3 + 4
avg_values2 = sum_values2 / 2
print(avg_values1)
print(avg_values2)


# Modular code
def avg_two_values(x, y):
    sum_values = x + y
    avg_values = sum_values / 2
    return avg_values


avg_values1 = avg_two_values(1, 2)
avg_values2 = avg_two_values(3, 4)
print(avg_values1)
print(avg_values2)
```
