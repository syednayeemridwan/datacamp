# Imperative vs Declarative Programming paradigms

- Imperative:
    - How the program should be executed
    - Requires step-by-step instructions
    - eg: Flowchart
    - Procedural and OOP programming
    - More common
- Declarative:
    - What the program should execute
    - Programmer defines logic instead of exact steps
    - eg: sudoku (rules are specified, but exact solution is not known)
    - Functional programming
    - Less common, not for general purpose

# Modular Code


- Code is broken into re-usable sections
- Avoid repeatition
- Separation of responsibilities
- Reduce chance of introducing bugs
- Saves development time
- Easier to understand

```python
# Non- Modular code
sum_values1 = 1 + 2
avg_values1 = sum_values1 / 2
sum_values2 = 3 + 4
avg_values2 = sum_values2 / 2
print(avg_values1)
print(avg_values2)


# Modular code
def avg_two_values(x, y):
    sum_values = x + y
    avg_values = sum_values / 2
    return avg_values


avg_values1 = avg_two_values(1, 2)
avg_values2 = avg_two_values(3, 4)
print(avg_values1)
print(avg_values2)
```

# Procedural Programming

- This paradigm uses procedures / subroutines
- A series of steps that can be re-run multiple times by one line of code 
- Aid with modularity (Modular Code)
- Helps with organization and readability
- commonly used in general purpose programming languages
- used in problems where each task can be broken into sub-tasks
- Great for data preparation processes
- Highly flexible
- NOTE: Procedural programming is imperative programming, but not all imperative programming is procedural
- Cons:
    - less secured (Process of how data is moved around is clearly visible)
    - Main focus is on operations on data rather than integrity of data in program
    - Rarely re-used across projects
- example: Functions in python

```python
def print_initial(name):
    initial = name[0]
    print(initial)


print_initial("Marwa")
print_initial("Celia")
print_initial("Raqael")
```

# Control Flow

- set of keywords and processes within a programming language that indicate how the logic of the code should be stepped through
- Examples of control flow statements:
    - if / else statements
    - for loops and while loops
    - Function definition using def


```python
def compare_heights(my_height, height_list):
    for height in height_list:
        if my_height > height:
            print("I am taller than ", height)
    return


my_height = 63
height_list = [62, 67, 70]
compare_heights(my_height, height_list)
```

# Pure functions

- a process that takes input values, produces output values based only on the input values, and does not do anything else
- Functional programming is a programming paradigm involving functions, specifically pure functions
- No influence/ Cannot modify other variables in the program outside the function scope (No side effects)
- pure functions are the mechanism by which functional programming employs separation of responsibilities.
- Non-pure function side effect example : Sorting the items within a list without creating a new variable isn't part of a pure function.
- Benefits:
    - Pure functions are easier to understand and debug
    - Testing of pure functions is easier
    - Output for a given input is entirely predictable
- Disadvantages:
    - Fewer experts, tools, frameworks exist for functional programming
    - Steeper learning curve and fewer educational resources
    - Larger memory usage limits applications


```python
def pure_sum(x, y):
    output = x + y
    return output


def not_pure_sum(x, y):
    output = x + y
    print(output)  # Side-effect
    return output
```

# From non-pure to pure Function

```python
# Not Pure Function
sample_mean = 10
scale_factor = 2


def scale_list(input_list):
    new_list = []
    for item in input_list:
        # Some variables are outside the function's scope
        new_item = (item - sample_mean) / scale_factor
        new_list.append(new_item)
    return new_list

# Pure Function


# Bringing variables inside function scope from outside
def scale_list(input_list, sample_mean, scale_factor):
    new_list = []
    for item in input_list:
        new_item = (item - sample_mean) / scale_factor
        new_list.append(new_item)
    return new_list
```

# Recursion

- Recursive function: a function that calls itself
- Must contain a termination condition, or base case
- Also contains the recursive call to itself with modified input

```python
def iterative_factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result


def recursive_factorial(n):
    if (n < 1):
        return 1
    else:
        return n*recursive_factorial(n-1)
```

# object-oriented programming

- Object-Oriented Programming: style of programming organized into "objects" 
- Objects: basic unit of object-oriented programming, can contain data and code
- Classes: categories of objects containing shared functionality and information among all objects from that class
- Neither classes nor objects are processes like functions, but can contain information about processes
- Class is General category of objects. eg: Dog
- Object is a specific example of a particular class. eg: Lacy the dog
- Public attributes and methods are available throughout the program (default in Python)
- Private attributes and methods are only accessible within the class itself (denoted with `__` prefix in Python)
- Class inheritance: when one class "inherits" methods and attributes from another, parent class
    - Example:
        - Poodle class inherits from Dog (and has curly hair)
        - Dog class inherits from Pet (and "barks")
        - Pet class inherits from Animal
- Advantages:
    - Problems where many small, similar tasks need to run at the same time (parrallalization)
    - Maintains security of data
    - Reusable and maintainable
    - Object-oriented databases (a specific type of database) like MongoDB
- Disadvantages:
    - Code can be slow to run
    - Programs can be longer
    - Not appropriate for all applications


### Simple class

```python
class Dog():
    def __init__(self, name):
        self.name = name

    def bark(self):
        print("Arf!")


lacy = Dog("Lacy")
lacy.bark()
```

### Public and Private Attributes

```python
class Dog():
    def __init__(self, name):
        self.name = name
        self.__hungry = True  # Private attribute

    def eat(self):
        self.__hungry = False


lacy = Dog("Lacy")
# This line won't work!! Because it is trying to access private attribute
lacy.__hungry = False
```

### Inheritance

```python
class Pet():
    def __init__(self, name):
        self.name = name


class Dog(Pet):
    def bark(self):
        print("Arf!")


class Cat(Pet):
    def meow(self):
        print("Meow!")


class Horse(Pet):
    def neigh(self):
        print("Neigh!")


lacy = Dog("Lacy")
lacy.bark()

fluffy = Cat("Fluffy")
fluffy.meow()

midnight = Horse("Midnight")
midnight.neigh()
```
