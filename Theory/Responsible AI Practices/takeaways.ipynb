# Responsible AI

- The practice of designing, developing and deploying AI with the awareness of ethical implications and societal impact
- Ensures that the AI systems are fair, transparent, accountable, while respecting user privacy and human rights.
- Also includes the value and principle of using AI systems
- Some biases:
    - recognising black and white people
    - discriminate based on age
    - lifting anonymiouty and exposing PII
- AI principles (by OECD):
    - Transparency and explainability
    - Fairness and non-discrimination
    - Robustness and Safety
    - Privacy and Data Governance
    - Accountability
    - Inclusive growth and sustainable development
- AI Ethics : philosophical question and ethical dilemmas
- RAI : Practical implementation of AI Ethics

# AI Risks

- Hallucinations : ChatGPT generating non-existant/inaccurate facts
- Data Poisoning and Corruption : Microsoft's chatbot became racist after user manipulation.
- Data breaches and leakage: some data gets leaked like medical research data
- Privacy risks: Privacy concerns
- Business risks:
    - Reputational Risks: AI bias to male candidates
    - Financial Risks: Misidentification of facial recognition and algorithmic bias

# AI Regulations

- EU : GDPR, DSA, EU AI Act
- US: Executive order on AI, American Data Privacy and Protection Act, Algorithmic Accountability Act
- Canada : Bill C-27 and Code of Conduct
- UK: Bletchly Declaration
- EU Risk levels:
    - Minimal Risk : That can be ignored, users can freely use AI.
    - Limited Risk : User interaction with AI. Users can make informed choices
    - High Risk : AI decisions in important sectors
    - Unacceptable Risk : Where AI practices pose clear threat to people's safety

# Strategy for Implementing Responsible AI

1. Embrace AI Governance : Highlight necessity of AI with business goal
2. AI Playbook: Integrating AI with existing business framework, well-documented
3. Identify Stakeholders : Understand their perspective to enhance AI solution
4. Leverage support mechanism : Guide and review Ethical issues to create awareness of AI risk and responsibilities
5. Embrace multi-stakeholder approach: Collaborate with communities to share insights and gain feedbacks
6. Additional Responsible Behavior Indicators: a company taking Certification of Responsible AI certification, which is a commitment
7. Implement governance tools: Outlines implementation of AI governance as a pyramid.
8. Monitor, Audit, Evaluate : Reviewing the whole system based on feedbacks iteratively.

# Maintaining Ethical Integrity

- Monitoring : Ongoing observation of AI systems
- Evaluating : Assessing the effectiveness and ethical performance against established criteria
    - Red and Blue teaming (Red team challenge the AI, Blue team respond to the red team evaluation)
    - Continuous evaluation : Makes sure both data and system remains reliable and quality-full, continuous scan for unfairness and robustness
- Auditing : Third-party review process that compliance with legal and regulatory requirements
