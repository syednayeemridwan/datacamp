# Responsible AI

- The practice of designing, developing and deploying AI with the awareness of ethical implications and societal impact
- Ensures that the AI systems are fair, transparent, accountable, while respecting user privacy and human rights.
- Also includes the value and principle of using AI systems
- Some biases:
    - recognising black and white people
    - discriminate based on age
    - lifting anonymiouty and exposing PII
- AI principles (by OECD):
    - Transparency and explainability
    - Fairness and non-discrimination
    - Robustness and Safety
    - Privacy and Data Governance
    - Accountability
    - Inclusive growth and sustainable development
- AI Ethics : philosophical question and ethical dilemmas
- RAI : Practical implementation of AI Ethics

# AI Threats

- AI for bad : Voice information, misinformation
- Workforce Changes : Impact on future job market
- AI Divide : Unequal opportunities and outcomes across nations
- Quantum computing : Too powerful and create security vulnerabilities
- Speed of AI Development : Regulations can never keep up with the speed of development of AI systems

# Tools for RAI

- Self-regulations : For everybody
- Frameworks
- Guidelines : Using existing ones in such a way so that the new ones can be kept under check
