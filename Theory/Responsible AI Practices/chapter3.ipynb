# Risk versus impact assessment

As part of your role at the tech company that employs AI in various products and services, you are involved in the development and oversight of AI models. Understanding the different types of assessments that these models undergo is crucial.

What is the primary difference between a risk assessment and an impact assessment in this context?

- Risk assessment focuses on identifying and mitigating potential negative outcomes, while impact assessment evaluates the broader effects of the AI model on stakeholders and the environment.
